### Base primária para RAG

Definiu-se que a base que será usada para o RAG seão de pedidos com avaliação diferente de NaN ou vazio, pois trazer NaN ou vazio para o fluxo não agregaria, todavia a nota dessas avaliações, bem como as outras informações serão importantes para o enriquecimento,

```python
df_reviews_rag = df[
    df["review_comment_message"].notna()
    & df["review_comment_message"].str.strip().ne("")
].copy()
```

```python
print("Total de registros:", len(df))
print("Registros selecionados:", len(df_reviews_rag))
print(
    "Pedidos únicos selecionados:",
    df_reviews_rag["order_id"].nunique()
)
print(
    "Reviews únicas selecionadas:",
    df_reviews_rag["review_id"].nunique()
)
```

```python
import pandas as pd

# Identificar avaliações válidas
df_analise = df.copy()

df_analise["comentario_valido"] = (
    df_analise["review_comment_message"].notna()
    & df_analise["review_comment_message"].str.strip().ne("")
)

# Resumir as avaliações por pedido
pedidos = (
    df_analise
    .groupby("order_id")
    .agg(
        qtd_avaliacoes=("review_id", "count"),
        qtd_validas=("comentario_valido", "sum")
    )
    .reset_index()
)

# Classificar os pedidos
pedidos["classificacao"] = "Mais de uma avaliação e não válidas"

pedidos.loc[
    (pedidos["qtd_avaliacoes"] == 1)
    & (pedidos["qtd_validas"] == 1),
    "classificacao"
] = "Uma avaliação válida"

pedidos.loc[
    (pedidos["qtd_avaliacoes"] == 1)
    & (pedidos["qtd_validas"] == 0),
    "classificacao"
] = "Uma avaliação não válida"

pedidos.loc[
    (pedidos["qtd_avaliacoes"] > 1)
    & (pedidos["qtd_validas"] > 0),
    "classificacao"
] = "Mais de uma avaliação, com válidas"

# Exibir o volume de pedidos por categoria
resultado = (
    pedidos["classificacao"]
    .value_counts()
    .reindex(
        [
            "Uma avaliação válida",
            "Uma avaliação não válida",
            "Mais de uma avaliação, com válidas",
            "Mais de uma avaliação e não válidas"
        ],
        fill_value=0
    )
    .rename_axis("categoria")
    .reset_index(name="volume_pedidos")
)

display(resultado)
```

```python
# Identificar pedidos com mais de uma linha
ids_duplicados = (
    df_reviews_rag["order_id"]
    .value_counts()
    .loc[lambda x: x > 1]
    .index
)

# Filtrar esses pedidos
df_pedidos_duplicados = df_reviews_rag[
    df_reviews_rag["order_id"].isin(ids_duplicados)
].copy()

# Contar pedidos com comentários iguais em todas as suas linhas
pedidos_com_comentarios_iguais = (
    df_pedidos_duplicados
    .groupby("order_id")["review_comment_message"]
    .nunique(dropna=False)
    .eq(1)
)

print(
    "Pedidos duplicados com comentários iguais:",
    pedidos_com_comentarios_iguais.sum()
)
```

```python
df_pedidos_duplicados[
    df_pedidos_duplicados["order_id"].isin(
        pedidos_com_comentarios_iguais[
            pedidos_com_comentarios_iguais
        ].index
    )
].sort_values("order_id")[
    ["order_id", "review_id", "review_score", "review_comment_message"]
]
```

```python
duplicados_comentario_nota = df_pedidos_duplicados[
    df_pedidos_duplicados.duplicated(
        subset=["order_id", "review_comment_message", "review_score"],
        keep=False
    )
].sort_values("order_id")


total_registros = len(duplicados_comentario_nota)

print(f"Total de registros: {total_registros}")

# Quantidade de pedidos afetados
total_pedidos = duplicados_comentario_nota["order_id"].nunique()

print(f"Total de pedidos: {total_pedidos}")
```
